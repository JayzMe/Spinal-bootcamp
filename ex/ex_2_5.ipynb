```scala
class T3 extends Component{
    val a = in SInt(8 bits)
    val b = out SInt()
    b := a.resize(16)
}
showRtl(new T3)
```

```scala
class T4 extends Component{
    val a = in SInt(8 bits)
    val b = out SInt(4 bits)
    b := a.resized
}
showRtl(new T4)
```

```scala
class T4 extends Component{
    val a = in SInt(8 bits)
    val b = out Bool()
    b := a.andR
}
showRtl(new T4)
```

```scala
class Pass extends Component{
    val a = in UInt()
    val b = out UInt()
    b := a
}
class Top extends Component{
    val x = in UInt(8 bits)
    val y = out UInt()

    val uut = new Pass
    uut.a := x
    y := uut.b
}
showRtl(new Top)
```

```scala
class T5 extends Component{
    val a0 = in SInt(8 bits)
    val a1 = in SInt(8 bits)
    val c = out SInt()

    c := (a0 ## a1).asSInt
}

showRtl(new T5)
```

```scala
class T6 extends Component{
    val a = in SInt(8 bits)
    val b = out SInt()
    val c = out SInt()

    val o0, o1, o2, o3 = out Bool()

    b := a(5 downto 2)
    c := a(0 to 4)
    o0 := a.msb
    o1 := a.lsb
    o2 := a(0)
    o3 := a(7)

    println(c.getWidth)
}
showRtl(new T6)
```
